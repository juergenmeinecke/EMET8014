## Background

In this assignment you will verify two foundational results from probability theory — the **Weak Law of Large Numbers (WLLN)** and the **Central Limit Theorem (CLT)** — through simulation.

We work with the **chi-squared distribution** with $k$ degrees of freedom, written $Y \sim \chi^2(k)$. Its key properties are:

- $E(Y) = k$
- $\text{Var}(Y) = 2k$

The chi-squared distribution is right-skewed, especially for small $k$. As $k$ grows, it becomes more symmetric. This makes it a useful test case for studying how distributional shape affects the CLT.

In Julia, create a chi-squared distribution via `Chisq(k)` from the `Distributions` package.

## Part III: The Central Limit Theorem

The CLT states that for a random sample $Y_1, \ldots, Y_N$ with $E(Y_i) = \mu$ and $\text{Var}(Y_i) = \sigma^2$:

$$\bar{Y}_N \overset{\text{approx}}{\sim} N\!\left(\mu,\; \frac{\sigma^2}{N}\right)$$

For $Y \sim \chi^2(k)$ this gives $\bar{Y}_N \overset{\text{approx}}{\sim} N(k,\; 2k/N)$.
