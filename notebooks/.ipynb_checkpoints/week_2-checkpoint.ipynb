## Part 3: Vectors in Julia

In econometrics, we work extensively with vectors (e.g., a column of observations) and matrices (e.g., design matrices). Let's learn how Julia handles these.

### Creating Vectors
Julia distinguishes between **column vectors** and **row vectors**. This matters for matrix operations!

```julia
# Three ways to create "vectors" - but they're not all the same!
y1 = [5, 8, 13]    # comma-separated
y2 = [5; 8; 13]    # semicolon-separated  
y3 = [5 8 13]      # space-separated

println("y1 = ", y1)
println("y2 = ", y2)
println("y3 = ", y3)
```

### Inner Products (Dot Products)
For vectors $\mathbf{a}$ and $\mathbf{b}$, the inner product is: $\mathbf{a} \cdot \mathbf{b} = \sum_i a_i b_i$

In Julia, you can compute this several ways:

```julia
a = [1, 2, 3]
b = [4, 5, 6]

# Method 1: Using the dot() function
@show dot(a, b)

# Method 2: Using transpose and multiplication
@show a' * b          # a' is the conjugate transpose of a

# Method 3: Using the Unicode dot operator (type \cdot then Tab)
@show a ⋅ b;
```

## Part 4: Broadcasting

**Broadcasting** is one of Julia's most powerful features. The `.` (dot) operator applies a function element-by-element.

This is essential in econometrics for operations like squaring residuals!

```julia
x = [1, 2, 3, 4, 5]

# Element-wise operations with broadcasting
@show x .* x          # element-wise multiplication (squaring)
@show x .^ 2          # another way to square
@show x .+ 10         # add 10 to each element
@show sqrt.(x)        # apply sqrt to each element
@show log.(x);        # apply log to each element
```

### Warning: Without the dot, things break!

```julia
# This will error - uncomment to see
# x * x    # Error! Can't multiply two column vectors
# x ^ 2    # Error! Can't square a vector
# sqrt(x)  # Error! sqrt expects a scalar
```

## Part 5: Matrices

### Creating Matrices
In Julia, matrices are created row by row. Spaces separate columns, semicolons separate rows.

```julia
# Create a 3×3 matrix
X = [1 2 3; 
     4 5 7; 
     8 9 11]

@show X
@show size(X)
@show typeof(X);
```

### Special Matrices
Julia provides convenient functions for common matrix types:

```julia
@show ones(3, 2)           # 3×2 matrix of ones
@show zeros(2, 4)          # 2×4 matrix of zeros
@show fill(3.14, 2, 2)     # 2×2 matrix filled with 3.14
@show I(3)                 # 3×3 identity matrix
@show Diagonal([1,2,3])    # diagonal matrix
@show rand(2, 3);          # 2×3 matrix of random numbers from U(0,1)
```

### Array Indexing and Slicing
Julia uses **1-based indexing** (like R and MATLAB, unlike Python).

```julia
X = [1 2 3; 4 5 7; 8 9 11]

@show X[1, 1]        # element at row 1, col 1
@show X[2, 3]        # element at row 2, col 3
@show X[:, 2]        # entire column 2
@show X[2, :]        # entire row 2  
@show X[1:2, 2:3]    # submatrix: rows 1-2, cols 2-3
@show X[end, :]      # last row
@show X[:, end];     # last column
```

## Part 6: Solving Linear Systems

A fundamental operation in econometrics is solving systems of linear equations. For OLS, we solve:
$$\mathbf{X}\boldsymbol{\beta} = \mathbf{y}$$

When $\mathbf{X}$ is square and invertible, the solution is $\boldsymbol{\beta} = \mathbf{X}^{-1}\mathbf{y}$.

```julia
X = [1 2 3; 4 5 7; 8 9 11]
y = [5, 8, 13]

# Check that X is invertible
@show rank(X)      # should equal 3 for a 3×3 invertible matrix
@show det(X);      # determinant should be non-zero
```

```julia
# Method 1: Explicit inverse (NOT recommended for numerical reasons)
β_inv = inv(X) * y
@show β_inv

# Method 2: Backslash operator (RECOMMENDED - more numerically stable)
β_backslash = X \ y
@show β_backslash

# Verify: X * β should equal y
@show X * β_backslash;
```

### Why use `\` instead of `inv()`?

1. **Numerical stability:** The backslash operator uses more stable algorithms
2. **Efficiency:** Computing the inverse is expensive and often unnecessary
3. **Generality:** `\` works for overdetermined systems (more equations than unknowns), which is the typical OLS case

**Rule of thumb:** If you're tempted to write `inv(A) * b`, use `A \ b` instead!

## Part 7: Important Gotcha - Aliasing

**Warning:** In Julia, assignment creates a *reference*, not a copy!

```julia
original = [1, 2, 3]
@show original

# This does NOT create a copy!
alias = original
alias[1] = 999

@show original    # original is also changed!
```

```julia
# To create an independent copy, use copy()
original = [1, 2, 3]
independent = copy(original)
independent[1] = 999

@show original     # unchanged!
@show independent  # changed
```

## Part 8: Useful Tips & Tricks

### Getting Help

```julia
# Use ? to access documentation
?sum
```

```julia
# Use methods() to see all available methods for a function
methods(sum)
```

### Ranges and Comprehensions

```julia
# Ranges
@show collect(1:5)           # integers 1 to 5
@show collect(0:0.5:2)       # 0, 0.5, 1.0, 1.5, 2.0
@show collect(range(0, 1, length=5));  # 5 evenly spaced points from 0 to 1
```

```julia
# List comprehensions (very Pythonic!)
@show [i^2 for i in 1:5]                    # squares
@show [i*j for i in 1:3, j in 1:3]          # multiplication table as matrix
@show [x for x in 1:10 if x % 2 == 0];      # even numbers (filtering)
```

### Chaining Operations with the Pipe Operator

```julia
# The |> operator passes output to the next function
data = [1, 2, 3, 4, 5]

# Calculate sqrt of sum of squares
result = data .|> (x -> x^2) |> sum |> sqrt
@show result

# Equivalent to:
@show sqrt(sum(data .^ 2));
```

| Concept | Key Syntax |
|---------|------------|
| Column vector | `[1, 2, 3]` or `[1; 2; 3]` |

| Broadcasting | `.` prefix: `x .^ 2`, `sin.(x)` |
| Inner product | `dot(a, b)` or `a' * b` |

| Indexing | 1-based: `X[1,1]`, `X[:, 2]`, `X[end, :]` |
| Copy arrays | `copy(x)` to avoid aliasing |
