## The Exponential Distribution

We will work with data that is **exponentially distributed**. The exponential distribution is commonly used in economics to model:
- Duration data (e.g., unemployment spells, time between transactions)
- Waiting times in queuing models
- Lifetimes of economic agents or products

The probability density function (pdf) of the exponential distribution is:

$$
f(y; \mu) = 
\begin{cases}
\frac{1}{\mu} e^{-y / \mu} & \text{if } y \geq 0 \\
0 & \text{if } y < 0
\end{cases}
$$

A random variable with this pdf is written as $Y \sim \text{Exp}(\mu)$.

**Key properties:**
- $E(Y) = \mu$ (the parameter is the mean)
- $\text{Var}(Y) = \mu^2$ (variance equals the square of the mean)
- The distribution is **memoryless**: $P(Y > s + t \mid Y > s) = P(Y > t)$

Notice that higher $\mu$ means both higher expected value *and* higher variance. This will be important when we study convergence rates.

## Exercise 3: Visualizing the Law of Large Numbers

### The Weak Law of Large Numbers (WLLN)

Given a random sample $Y_1, \ldots, Y_N$ of i.i.d. random variables with $E(Y_i) = \mu$, the WLLN states:

$$\bar{Y}_N \xrightarrow{p} \mu \quad \text{as } N \to \infty$$

In words: the sample mean **converges in probability** to the population mean as the sample size grows.

**Economic significance:** This theorem is fundamental to econometrics. It tells us that:
1. Sample averages are consistent estimators of population means
2. With enough data, our estimates will be arbitrarily close to the true values
3. This underpins the consistency of many econometric estimators (OLS, IV, GMM, etc.)

### Implementation Strategy

To visualize the WLLN, we:
1. Draw a large "supersample" of size $n_{\max}$
2. For each $n = 1, 2, \ldots, n_{\max}$, compute $\bar{Y}_n$ using the first $n$ observations
3. Plot $\bar{Y}_n$ against $n$ and observe convergence to $\mu$

### How Variance Affects Convergence Speed

The WLLN tells us *that* convergence occurs, but not *how fast*. The speed of convergence depends on the variance of the underlying distribution.

For the exponential distribution, $\text{Var}(Y) = \mu^2$, so:
- **Smaller $\mu$** → smaller variance → faster convergence
- **Larger $\mu$** → larger variance → slower convergence

**Intuition:** With higher variance, individual observations are more "noisy," so you need more data to average out the noise and get close to the true mean.
